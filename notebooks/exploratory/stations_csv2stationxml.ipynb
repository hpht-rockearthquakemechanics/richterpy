# Convert Station CSV To StationXML

Exploratory notebook for the station CSV metadata path: `A2 -> O1 -> O`.

```python
from pathlib import Path

from richterpy.convert.stations import convert_stations

repo_root = Path.cwd()
for candidate in (repo_root, *repo_root.parents):
    if (candidate / 'config' / 'MEERA.ini').exists():
        repo_root = candidate
        break
else:
    raise FileNotFoundError('Could not locate repo root from current working directory.')

experiment = 'm0013'
datum_config = repo_root / 'config' / 'MEERA.ini'
csv_path = repo_root / 'data' / experiment / 'sensorarray' / f'{experiment}.csv'
output_path = repo_root / 'data' / 'playground' / f'{experiment}.stations.csv.xml'
output_path.parent.mkdir(parents=True, exist_ok=True)
```

```python
convert_stations(
    experiment_id=experiment,
    csv_path=csv_path,
    output_path=output_path,
    datum_config=datum_config,
)
```
