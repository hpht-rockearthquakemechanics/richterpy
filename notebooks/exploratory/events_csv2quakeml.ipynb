# Convert Event CSV To QuakeML

Exploratory notebook for the CSV event metadata path: `C1 -> O4 -> O`.

```python
from pathlib import Path

from richterpy.convert.events import convert_events

repo_root = Path.cwd()
for candidate in (repo_root, *repo_root.parents):
    if (candidate / 'config' / 'MEERA.ini').exists():
        repo_root = candidate
        break
else:
    raise FileNotFoundError('Could not locate repo root from current working directory.')

experiment = 'm0013'
datum_config = repo_root / 'config' / 'MEERA.ini'
csv_path = repo_root / 'data' / experiment / 'export' / f'{experiment} event data.csv'
output_path = repo_root / 'data' / 'playground' / f'{experiment}.events.csv.xml'
output_path.parent.mkdir(parents=True, exist_ok=True)
```

```python
convert_events(
    experiment_id=experiment,
    csv_path=csv_path,
    output_path=output_path,
    datum_config=datum_config,
)
```
